```python
import pandas as pd
import matplotlib.pyplot as plt
from utils import kplotlib as kpl

kpl.init_kplotlib()
```

```python
file_path = r"C:\Users\Saroj Chand\Downloads\Davis_118 Trend.csv"
df = pd.read_csv(file_path)
```

```python
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
df = df.sort_values("Timestamp")

# 15-minute rolling mean for 15-minute sampled data
df["Temp_rolling_15m"] = df["Lab118SpaceTemp"].rolling(window=1, center=True).mean()

plt.figure(figsize=(11, 5))
plt.plot(df["Timestamp"], df["Lab118SpaceTemp"], ".", alpha=0.7, label="Raw")
plt.plot(
    df["Timestamp"],
    df["Temp_rolling_15m"],
    "-",
    linewidth=2,
    label="15-minute rolling mean",
)
plt.xlabel("Time")
plt.ylabel("Lab118SpaceTemp (°F)")
plt.title("Lab 118 Space Temperature")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
